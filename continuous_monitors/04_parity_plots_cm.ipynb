```python
import matplotlib.pyplot as plt
from plot_methods_cm import make_parity_plot, get_parity_data
```

```python
cm_ops = ['SLB', 'Sensia', 'Sensirion']
# Initialize figure
fig, ax = plt.subplots(1, 3, figsize=[18, 35])
ax = ax.flatten()

for a, operator in enumerate(cm_ops):
    # Generate parity data
    parity_data, parity_notes = get_parity_data(operator=operator)

    # Make figure
    ax[a] = make_parity_plot(parity_data, parity_notes, ax[a], plot_lim='largest_kgh')

    # Axes labels
    ax[a].set_xlabel('Methane Release Rate (kgh)', fontsize=11)
    ax[a].set_ylabel('Reported Release Rate (kgh)', fontsize=11)
```

```python
fig, ax = plt.subplots(1, 1, figsize=[6, 6])
parity_data, parity_notes = get_parity_data(operator='all', operator_list=['Sensia', 'Sensirion', 'SLB'])
# Make figure
ax = make_parity_plot(parity_data, parity_notes, ax, plot_lim='largest_kgh')
# Axes labels
ax.set_xlabel('Methane Release Rate (kgh)', fontsize=11)
ax.set_ylabel('Reported Release Rate (kgh)', fontsize=11)
```
