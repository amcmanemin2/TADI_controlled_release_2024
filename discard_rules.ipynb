```python
import pandas as pd
from methods import generate_release_summary
```

```python
def print_qc_summary(op):
    print(f'OPERATOR: {op}')
    df = generate_release_summary(op)

    ## STRICT QC
    true_counts = df.loc[df["strict_qc_kept"], "operator_quantification"].agg(
        zeros=lambda x: (x == 0).sum(),
        NAs=lambda x: x.isna().sum(),
        non_zeros=lambda x: ((x != 0) & x.notna()).sum()
    )
    print('Strict QC:')
    print('     Total count: ', len(df))
    print(f'     False count: {(df["strict_qc_kept"] == False).sum()}')
    print(f'     True counts: {true_counts["zeros"]} zeros, {true_counts["non_zeros"]} non-zeros, {true_counts["NAs"]} NAs, {(df["strict_qc_kept"] == True).sum()} total \n')

    ## STANFORD QC
    true_counts = df.loc[df["stanford_kept"], "operator_quantification"].agg(
        zeros=lambda x: (x == 0).sum(),
        NAs=lambda x: x.isna().sum(),
        non_zeros=lambda x: ((x != 0) & x.notna()).sum()
    )
    print('Stanford QC:')
    print('     Total count: ', len(df))
    print(f'     False count: {(df["stanford_kept"] == False).sum()}')
    print(f'     True counts: {true_counts["zeros"]} zeros, {true_counts["non_zeros"]} non-zeros, {true_counts["NAs"]} NAs, {(df["stanford_kept"] == True).sum()} total \n')

    ## OPERATOR QC
    true_counts = df.loc[df["operator_kept"], "operator_quantification"].agg(
        zeros=lambda x: (x == 0).sum(),
        NAs=lambda x: x.isna().sum(),
        non_zeros=lambda x: ((x != 0) & x.notna()).sum()
    )
    print('Operator QC:')
    print('     Total count: ', len(df))
    print(f'     False count: {(df["operator_kept"] == False).sum()}')
    print(f'     True counts: {true_counts["zeros"]} zeros, {true_counts["non_zeros"]} non-zeros, {true_counts["NAs"]} NAs, {(df["operator_kept"] == True).sum()} total \n')

    return df
```

```python
op = 'Sensirion'
df = print_qc_summary(op)
```
