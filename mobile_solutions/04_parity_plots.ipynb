## Drone Teams

```python
import matplotlib.pyplot as plt
from plot_methods import make_parity_plot, get_parity_data
```

```python
drone_ops = ['Aeromon', 'SeekOps', 'GSMA', 'Flylogix']
# Initialize figure
fig, ax = plt.subplots(2, 2, figsize=[12, 12])
ax = ax.flatten()

for a, operator in enumerate(drone_ops):
    # Generate parity data
    parity_data, parity_notes = get_parity_data(operator=operator)

    # Make figure
    ax[a] = make_parity_plot(parity_data, parity_notes, ax[a], plot_lim='largest_kgh')

    # Axes labels
    ax[a].set_xlabel('Methane Release Rate (kgh)', fontsize=11)
    ax[a].set_ylabel('Reported Release Rate (kgh)', fontsize=11)
```

## Vehicle Teams

```python
vehicle_ops = ['DTU', 'UHEI']
# Initialize figure
fig, ax = plt.subplots(1, 2, figsize=[12, 12])
ax = ax.flatten()

for a, operator in enumerate(vehicle_ops):
    # Generate parity data
    parity_data, parity_notes = get_parity_data(operator=operator)

    # Make figure
    ax[a] = make_parity_plot(parity_data, parity_notes, ax[a], plot_lim='largest_kgh')

    # Axes labels
    ax[a].set_xlabel('Methane Release Rate (kgh)', fontsize=11)
    ax[a].set_ylabel('Reported Release Rate (kgh)', fontsize=11)
```
